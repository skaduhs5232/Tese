# `00_join.ipynb` — União dos 8 Tiles Globais do GEBCO 2026 Sub-Ice

Une os 8 recortes (`datasets/raw/gebco_2026_sub_ice_*.tif`) em um único mosaico global `datasets/processed/gebco_2026_sub_ice_geotiff.tif`, detectando automaticamente as dimensões dos tiles.

```python
import numpy as np
from geotiff.geotiff import ReadGeoTiff, WriteGeoTiff
```

```python
filenames = [[r'datasets/raw/gebco_2026_sub_ice_n90.0_s0.0_w-180.0_e-90.0_geotiff.tif',
              r'datasets/raw/gebco_2026_sub_ice_n90.0_s0.0_w-90.0_e0.0_geotiff.tif',
              r'datasets/raw/gebco_2026_sub_ice_n90.0_s0.0_w0.0_e90.0_geotiff.tif',
              r'datasets/raw/gebco_2026_sub_ice_n90.0_s0.0_w90.0_e180.0_geotiff.tif'],
             [r'datasets/raw/gebco_2026_sub_ice_n0.0_s-90.0_w-180.0_e-90.0_geotiff.tif',
              r'datasets/raw/gebco_2026_sub_ice_n0.0_s-90.0_w-90.0_e0.0_geotiff.tif',
              r'datasets/raw/gebco_2026_sub_ice_n0.0_s-90.0_w0.0_e90.0_geotiff.tif',
              r'datasets/raw/gebco_2026_sub_ice_n0.0_s-90.0_w90.0_e180.0_geotiff.tif']]

info0, arr0 = ReadGeoTiff(filenames[0][0])
height_tile, width_tile = arr0.shape
height = height_tile * 2
width = width_tile * 4

arr = np.empty((height, width), dtype=np.int16)
info = {
    'RasterYSize': height,
    'RasterXSize': width,
    'GeoTransform': info0['GeoTransform'],
    'Projection': info0['Projection'],
    'NoDataValue': info0['NoDataValue'],
}

for j, row in enumerate(filenames):
    for i, filename in enumerate(row):
        _info, _arr = ReadGeoTiff(filename)
        j_min, j_max = j * height_tile, (j + 1) * height_tile
        i_min, i_max = i * width_tile, (i + 1) * width_tile
        arr[j_min:j_max, i_min:i_max] = _arr

print(f'Mosaico global construído: {width} x {height} pixels')
```

```python
filename = 'datasets/processed/gebco_2026_sub_ice_geotiff.tif'
WriteGeoTiff(info, arr, filename)
print(f'Salvo com sucesso em {filename}')
```
